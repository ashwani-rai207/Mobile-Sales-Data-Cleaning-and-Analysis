```python

import pandas as pd
from IPython.display import display
from google.colab import files

# ==========================================
# CUSTOMER DATA CLEANING & ANALYSIS PROJECT
# INPUT + FIVE OUTPUTS (WITHOUT GRAPHS)
# ==========================================

# FIXED FILE PATH - NO CHANGE
file_path = "/content/customer_1.csv"

# ==========================================
# STEP 1: LOAD DATASET
# ==========================================

df = pd.read_csv(
    file_path,
    dtype={"Phone": "string"}
)

df.columns = df.columns.str.strip()

print("DATASET LOADED SUCCESSFULLY!")

# ==========================================
# STEP 2: TAKE USER INPUT
# ==========================================

print("\n========== USER INPUT ==========")

print("1. Search By City")
print("2. Search By State")
print("3. Search By Phone Brand")
print("4. Search By Operating System")
print("5. Analyze All Customers")

choice = input("\nEnter Your Choice (1-5): ").strip()

# ==========================================
# OUTPUT 1: ORIGINAL DATASET
# ==========================================

print("\n========== OUTPUT 1 ==========")
print("ORIGINAL CUSTOMER DATASET")

display(df.head(10))

print("\nTotal Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

# ==========================================
# STEP 3: DATA CLEANING
# ==========================================

clean_df = df.copy()

# Count duplicate rows
duplicate_rows = clean_df.duplicated().sum()

# Remove duplicate rows
clean_df.drop_duplicates(inplace=True)

# Remove duplicate Customer IDs
clean_df.drop_duplicates(
    subset=["CustomerID"],
    keep="first",
    inplace=True
)

clean_df.reset_index(drop=True, inplace=True)

# Text columns
text_columns = [
    "Name",
    "City",
    "State",
    "Phone Brand",
    "Operating System"
]

# Remove extra spaces
for col in text_columns:
    clean_df[col] = (
        clean_df[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# Standardize names and locations
clean_df["Name"] = clean_df["Name"].str.title()
clean_df["City"] = clean_df["City"].str.title()
clean_df["State"] = clean_df["State"].str.title()

# Standardize phone brands
clean_df["Phone Brand"] = (
    clean_df["Phone Brand"].str.title()
)

# Standardize operating systems
clean_df["Operating System"] = (
    clean_df["Operating System"]
    .str.lower()
    .replace({
        "ios": "iOS",
        "android": "Android"
    })
)

# Fill missing text values
for col in text_columns:
    clean_df[col] = clean_df[col].fillna("Unknown")

# ==========================================
# STEP 4: PHONE NUMBER CLEANING
# ==========================================

# Preserve original phone number
clean_df["Original_Phone"] = clean_df["Phone"]

# Remove spaces and special characters
clean_df["Phone"] = (
    clean_df["Phone"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
)

# Remove Indian country code
clean_df["Phone"] = (
    clean_df["Phone"]
    .str.replace(
        r"^91(?=[6-9]\d{9}$)",
        "",
        regex=True
    )
)

# Validate phone numbers
valid_phone = (
    clean_df["Phone"]
    .str.fullmatch(r"[6-9]\d{9}")
    .fillna(False)
)

clean_df["Phone_Status"] = valid_phone.map({
    True: "Valid",
    False: "Invalid"
})

# ==========================================
# OUTPUT 2: DATA CLEANING REPORT
# ==========================================

print("\n========== OUTPUT 2 ==========")
print("DATA CLEANING REPORT")

print("\nOriginal Records:", len(df))
print("Cleaned Records:", len(clean_df))

print("\nDuplicate Rows Found:", duplicate_rows)

print(
    "Total Records Removed:",
    len(df) - len(clean_df)
)

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

print("\nMissing Values After Cleaning:")
print(clean_df.isnull().sum())

print("\nPhone Number Status:")
print(clean_df["Phone_Status"].value_counts())

# ==========================================
# OUTPUT 3: CLEANED DATASET
# ==========================================

print("\n========== OUTPUT 3 ==========")
print("CLEANED CUSTOMER DATASET")

display(clean_df.head(10))

print("\nCleaned Dataset Shape:")
print(clean_df.shape)

# Save cleaned CSV
output_path = "/content/customer_project_cleaned.csv"

clean_df.to_csv(
    output_path,
    index=False
)

print("\nCleaned File Saved Successfully!")
print("File Location:", output_path)

# ==========================================
# STEP 5: FILTER USING USER INPUT
# ==========================================

filtered_df = clean_df.copy()

if choice in ["1", "2", "3", "4"]:

    columns = {
        "1": "City",
        "2": "State",
        "3": "Phone Brand",
        "4": "Operating System"
    }

    selected_column = columns[choice]

    print("\nAvailable Values:")
    print(clean_df[selected_column].value_counts())

    user_value = input(
        "\nEnter " + selected_column + ": "
    ).strip()

    filtered_df = clean_df[
        clean_df[selected_column].str.casefold()
        == user_value.casefold()
    ].copy()

elif choice == "5":
    print("\nAnalyzing All Customers")

else:
    print("\nInvalid Choice! Analyzing All Customers.")

# ==========================================
# OUTPUT 4: CUSTOMER SEARCH RESULT
# ==========================================

print("\n========== OUTPUT 4 ==========")
print("CUSTOMER SEARCH RESULT")

print("\nMatching Customers:", len(filtered_df))

if filtered_df.empty:
    print("No Matching Customer Found!")

else:
    display(filtered_df.head(20))

# ==========================================
# OUTPUT 5: CUSTOMER DATA ANALYSIS
# ==========================================

print("\n========== OUTPUT 5 ==========")
print("CUSTOMER DATA ANALYSIS")

if filtered_df.empty:
    print("No Data Available For Analysis")

else:
    print("\nTotal Matching Customers:", len(filtered_df))

    print("\nCustomers By City:")
    print(filtered_df["City"].value_counts())

    print("\nCustomers By State:")
    print(filtered_df["State"].value_counts())

    print("\nPhone Brand Distribution:")
    print(filtered_df["Phone Brand"].value_counts())

    print("\nOperating System Distribution:")
    print(filtered_df["Operating System"].value_counts())

    print("\nPhone Number Validation:")
    print(filtered_df["Phone_Status"].value_counts())

# ==========================================
# DOWNLOAD CLEANED CSV
# ==========================================

print("\n===================================")
print("PROJECT COMPLETED SUCCESSFULLY!")
print("===================================")

print("\nCleaned Dataset Saved At:")
print(output_path)

files.download(output_path)
```
